````python
actor_jahre = vereine[
    (vereine['Foerderung_ref'] == 'MA7') & 
    (vereine['Typ'] == 'Beschluss')
].groupby('Verein')['Jahr'].apply(set).reset_index()

actor_jahre.columns = ['Verein', 'Jahre']

actor_jahre['n_jahre'] = actor_jahre['Jahre'].apply(len)
actor_jahre['2022'] = actor_jahre['Jahre'].apply(lambda x: '2022' in x)
actor_jahre['2023'] = actor_jahre['Jahre'].apply(lambda x: '2023' in x)
actor_jahre['2024'] = actor_jahre['Jahre'].apply(lambda x: '2024' in x)
actor_jahre['kontinuierlich_2224'] = actor_jahre['Jahre'].apply(
    lambda x: {'2022','2023','2024'}.issubset(x)
)

print(f"total actores: {len(actor_jahre)}")
print(f"con beschluss en 2022+2023+2024: {actor_jahre['kontinuierlich_2224'].sum()}")
print(f"\nactores continuos:")
print(actor_jahre[actor_jahre['kontinuierlich_2224']][['Verein', 'Jahre']].to_string())
````
